```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn import svm
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import VotingClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df = pd.read_csv('CC_Datasets.csv')
df.head()
```

```python
def apply_response(label):
    return 0 if label == 0 else 1

df['results'] = df['is_fraud'].apply(apply_response)
df[['trans_num', 'is_fraud', 'results']].head()
```

```python
cv = CountVectorizer()
X = df['trans_num']
y = df['results']

print("Transaction Number")
print(X.head())

print("Results")
print(y.head())
```

```python
X = cv.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

print("Train/Test Shape:")
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
print("Naive Bayes")
NB = MultinomialNB()
NB.fit(X_train, y_train)
predict_nb = NB.predict(X_test)

print("ACCURACY:", accuracy_score(y_test, predict_nb) * 100)
print("CLASSIFICATION REPORT\n", classification_report(y_test, predict_nb))
print("CONFUSION MATRIX\n", confusion_matrix(y_test, predict_nb))
```

```python
print("SVM")
lin_clf = svm.LinearSVC()
lin_clf.fit(X_train, y_train)
predict_svm = lin_clf.predict(X_test)

print("ACCURACY:", accuracy_score(y_test, predict_svm) * 100)
print("CLASSIFICATION REPORT\n", classification_report(y_test, predict_svm))
print("CONFUSION MATRIX\n", confusion_matrix(y_test, predict_svm))
```

```python
print("Logistic Regression")
reg = LogisticRegression(random_state=0, solver='lbfgs')
reg.fit(X_train, y_train)
y_pred = reg.predict(X_test)

print("ACCURACY:", accuracy_score(y_test, y_pred) * 100)
print("CLASSIFICATION REPORT\n", classification_report(y_test, y_pred))
print("CONFUSION MATRIX\n", confusion_matrix(y_test, y_pred))
```

```python
models = [
    ('naive_bayes', NB),
    ('svm', lin_clf),
    ('logistic', reg)
]

classifier = VotingClassifier(models)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)

print("Voting Classifier Accuracy:", accuracy_score(y_test, y_pred) * 100)
```

```python

trans_num = "C1234567890"

trans_num1 = [trans_num]
vector1 = cv.transform(trans_num1)
predict_text = classifier.predict(vector1)

prediction = int(predict_text[0])
val = 'No Credit Card Fraud' if prediction == 0 else 'Credit Card Fraud'

print("Prediction:", val)
```
